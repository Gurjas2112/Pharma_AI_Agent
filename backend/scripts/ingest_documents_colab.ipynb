## Step 4: Initialize Embedding Service

```python
from llama_cpp import Llama
import os

print("Loading Mistral model for embeddings...")
print(f"Model path: {MODEL_PATH}")
print(f"File exists: {os.path.exists(MODEL_PATH)}")

# Initialize Mistral for embeddings
model = Llama(
    model_path=MODEL_PATH,
    embedding=True,
    n_ctx=2048,
    n_threads=2,  # Colab has 2 CPUs
    verbose=False
)

print("✓ Mistral model loaded successfully")

async def generate_embedding(text: str):
    """Generate 1536-dim embedding"""
    embedding = model.embed(text)
    
    # Pad or truncate to 1536 dimensions
    if len(embedding) < 1536:
        embedding.extend([0.0] * (1536 - len(embedding)))
    elif len(embedding) > 1536:
        embedding = embedding[:1536]
    
    return embedding

# Test
import asyncio
test_emb = await generate_embedding("test")
print(f"✓ Embedding dimension: {len(test_emb)}")
```

## Step 6: Document Ingestion Functions

```python
import pdfplumber
import re
from pathlib import Path
from typing import List, Dict
from tqdm.notebook import tqdm

def extract_text_from_pdf(pdf_path: Path) -> str:
    """Extract text from PDF"""
    text_content = []
    
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                text_content.append(text)
    
    return "\n\n".join(text_content)

def chunk_text(text: str, source: str, chunk_size: int = 1000, overlap: int = 200) -> List[Dict]:
    """Split text into overlapping chunks"""
    chunks = []
    text = re.sub(r'\s+', ' ', text).strip()
    
    start = 0
    chunk_index = 0
    
    while start < len(text):
        end = start + chunk_size
        chunk_text = text[start:end]
        
        # Try to break at sentence boundary
        if end < len(text):
            last_period = chunk_text.rfind('. ')
            if last_period > chunk_size * 0.5:
                chunk_text = chunk_text[:last_period + 1]
                end = start + last_period + 1
        
        chunks.append({
            "source": source,
            "content": chunk_text.strip(),
            "metadata": {
                "chunk_index": chunk_index,
                "chunk_start": start,
                "chunk_end": end
            }
        })
        
        chunk_index += 1
        start = end - overlap
    
    # Update total_chunks
    for chunk in chunks:
        chunk["metadata"]["total_chunks"] = len(chunks)
    
    return chunks

async def insert_chunk(chunk: Dict):
    """Insert chunk into Supabase"""
    try:
        result = supabase.table("internal_docs").insert({
            "source": chunk["source"],
            "content": chunk["content"],
            "embedding": chunk["embedding"],
            "metadata": chunk["metadata"]
        }).execute()
        return True
    except Exception as e:
        print(f"Error inserting chunk: {e}")
        return False

print("✓ Functions defined")
```

## Step 7: Process All PDFs

**This will take 30-60 minutes depending on PDF sizes**

```python
import asyncio
from pathlib import Path

async def ingest_all_pdfs():
    pdf_files = list(Path(PDF_DIRECTORY).glob("*.pdf"))
    
    if not pdf_files:
        print(f"✗ No PDFs found in {PDF_DIRECTORY}")
        return
    
    print(f"Found {len(pdf_files)} PDF files")
    print("="*60)
    
    total_chunks_inserted = 0
    
    for pdf_path in tqdm(pdf_files, desc="Processing PDFs"):
        print(f"\n📄 {pdf_path.name}")
        
        # Extract text
        print("  Extracting text...")
        text = extract_text_from_pdf(pdf_path)
        print(f"  ✓ Extracted {len(text):,} characters")
        
        if not text:
            print("  ✗ No text extracted, skipping")
            continue
        
        # Chunk text
        print("  Chunking text...")
        chunks = chunk_text(text, pdf_path.name)
        print(f"  ✓ Created {len(chunks)} chunks")
        
        # Generate embeddings and insert
        print("  Generating embeddings and inserting...")
        success_count = 0
        
        for i, chunk in enumerate(tqdm(chunks, desc=f"  Chunks", leave=False)):
            # Generate embedding
            embedding = await generate_embedding(chunk["content"])
            chunk["embedding"] = embedding
            
            # Insert to Supabase
            if await insert_chunk(chunk):
                success_count += 1
        
        print(f"  ✓ Inserted {success_count}/{len(chunks)} chunks")
        total_chunks_inserted += success_count
    
    print("\n" + "="*60)
    print(f"✓ Ingestion complete: {total_chunks_inserted} total chunks inserted")
    print("="*60)

# Run ingestion
await ingest_all_pdfs()
```
